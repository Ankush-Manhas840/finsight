```python
from pypdf import PdfReader

reader = PdfReader("../data/raw/mahindra_FY2025.pdf")
print("Total pages:", len(reader.pages))

for page_number, page in enumerate(reader.pages, start=1):
    text = page.extract_text() or ""
    if "standalone statement of profit and loss" in text.lower():
        print("Found on PDF page", page_number)
```

```python
page_text = reader.pages[146 - 1].extract_text()   # change 175 to the page you found

keywords = ["revenue from operations", "profit for the year",
            "borrowings", "total equity", "earnings per"]

for line in page_text.splitlines():
    if any(word in line.lower() for word in keywords):
        print(line)
```
