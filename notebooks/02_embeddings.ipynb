```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

sentences = [
    "Net profit increased this year",
    "PAT grew compared to last year",
    "The company sold more cars",
    "India won the cricket match",
]

embeddings = model.encode(sentences)
print(embeddings.shape)

scores = model.similarity(embeddings, embeddings)
print(scores.round(decimals=2))
```

```python
short_forms = {
    "pat": "profit after tax",
    "eps": "earnings per share",
    "roe": "return on equity",
}

def expand(sentence):
    sentence=sentence.lower()
    # your code here
    # hint 1: sentence.split() gives you a list of words
    # hint 2: loop through the words; if word.upper() is in short_forms, swap it
    # hint 3: " ".join(list_of_words) turns the list back into a sentence
    checker=sentence.split()
    result=[]
    for i in checker:
        if i in short_forms:
            i=short_forms[i]
        result.append(i)
    return " ".join(result)
   

print(expand("PAT grew compared to last year"))
print(expand("The patent was filed"))
```

```python
question = "What was the PAT this year?"
report_lines = [
    "Profit for the year increased to 11,855 crore",
    "The company sold more cars this year",
    "Earnings per share was 98.80 rupees",
    "India won the cricket match",
]

for q in [question, expand(question)]:
    q_emb = model.encode(q)
    lines_emb = model.encode(report_lines)
    scores = model.similarity(q_emb, lines_emb)
    print("Question:", q)
    print(scores.round(decimals=2))
    print()
```
